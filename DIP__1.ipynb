## Q1. Many scientific and engineering images are not stored as uint8 but as other data types. How can these images be viewed?

### Answer:

Most images are stored as uint8 where pixel values range from 0 to 255.

However scientific and engineering images are often stored as:

- uint16
- float32
- float64

These images cannot be displayed correctly directly because their values exceed the normal 0–255 range.

To display them:

1. Normalize pixel values
2. Convert them to uint8
3. Display using Matplotlib or OpenCV

Normalization scales values between 0 and 255.

Formula:

Normalized Image =
(Image - Minimum Value)
/
(Maximum Value - Minimum Value)

This allows proper visualization of scientific images.

```python
import numpy as np
import matplotlib.pyplot as plt

img = np.random.rand(256,256)*5000

normalized = (img-img.min())/(img.max()-img.min())

plt.imshow(normalized,cmap='gray')
plt.title("Normalized Scientific Image")
plt.show()
```

## Q4. Image Hiding using Steganography

Image Steganography is the process of hiding information inside another image.

In this implementation:

- Cover Image stores the secret image.
- Secret Image is hidden inside the cover image.
- Least Significant Bit (LSB) concept is used.

The hidden image can later be extracted.

```python
from PIL import Image
import numpy as np

cover = Image.open("cover.jfif")
secret = Image.open("secret.jfif")

cover = cover.resize((300,300))
secret = secret.resize((300,300))

cover = np.array(cover)
secret = np.array(secret)

encoded = cover.copy()

encoded[:,:,0] = (cover[:,:,0] & 240) | (secret[:,:,0] >> 4)
encoded[:,:,1] = (cover[:,:,1] & 240) | (secret[:,:,1] >> 4)
encoded[:,:,2] = (cover[:,:,2] & 240) | (secret[:,:,2] >> 4)

Image.fromarray(encoded).save("encoded.png")

print("Secret image hidden successfully")
```

##Extracting hidden image.

```python
from PIL import Image
import numpy as np

encoded = np.array(Image.open("encoded.png"))

decoded = np.zeros_like(encoded)

decoded[:,:,0] = (encoded[:,:,0] & 15) << 4
decoded[:,:,1] = (encoded[:,:,1] & 15) << 4
decoded[:,:,2] = (encoded[:,:,2] & 15) << 4

Image.fromarray(decoded).save("decoded.png")

print("Hidden image extracted")
```
